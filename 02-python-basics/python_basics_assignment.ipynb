### 27. Filter messages

Create a new list of non-blank messages, trimmed of surrounding spaces.

### 28. Count passing scores

Count scores greater than or equal to 50. Also try an empty list and confirm the result is zero.

**Sample expected output:** `2`

```python
scores = [20, 50, 80, 35]
# Write your solution below.
def count_passing(score_list):
    count = 0
    for s in score_list:
        if s >= 50:
            count += 1
    return count

print(count_passing(scores))
print(count_passing([]))  # empty list -> 0
```

### 30. Mini project: support messages

Trim each message. Count blank messages, non-blank messages longer than 10 characters, and other ready messages. Print all three counts. A message belongs to only one category.

**Sample expected output:** `ready: 1, blank: 1, too long: 1`

```python
messages = [" Hi ", " ", "Where is my order?"]
# Write your solution below.
ready = 0
blank = 0
too_long = 0
for m in messages:
    m = m.strip()
    if m == "":
        blank += 1
    elif len(m) > 10:
        too_long += 1
    else:
        ready += 1
print(f"ready: {ready}, blank: {blank}, too long: {too_long}")
```
